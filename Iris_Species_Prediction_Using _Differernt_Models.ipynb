```python
import pandas as pd 
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import accuracy_score, precision_score, recall_score
from sklearn.metrics import confusion_matrix, classification_report, f1_score
```

```python
iris_data=pd.read_csv("Iris.csv")
```

```python
iris_data.head()
```

```python
iris_data.isnull().sum()
```

```python
iris_data.describe()
```

```python
# convert 3 species into 0, 1, 2
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
iris_data['Species'] = le.fit_transform(iris_data['Species'])
```

```python
x= iris_data.drop(["Id","Species"], axis=1)
y= iris_data["Species"]
```

```python
x.sample()
```

```python
y.sample()
```

```python
# Classification-Model --->  Logistic Regrssion

x_train,x_test,y_train,y_test= train_test_split(
    x, y, test_size=0.5, random_state=42
)

model1= LogisticRegression(max_iter=1000)

model1.fit(x_train,y_train)

y_predict= model1.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_predict)*100, "%")

print("Precision:",  precision_score(y_test, y_predict, average='macro')*100, "%")

print("Recall:", recall_score(y_test, y_predict, average='macro')*100, "%")

print("f1_score:", f1_score(y_test, y_predict, average='macro')*100 , "%")

print("Confusion-Matrix:", confusion_matrix(y_test,y_predict))
result = pd.DataFrame({
    "Actual": y_test,
    "Prediction": y_predict
})

print(result)

print("Classification-Report:", classification_report(y_test, y_predict)) # return string as output
```

```python
# Classification-Model --->  Naive_Bayes (Gaussian)

GNB_model= GaussianNB()

GNB_model.fit(x_train, y_train)

y_pred= GNB_model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred)*100, "%")

print("Precision:",  precision_score(y_test, y_pred, average='macro')*100, "%")

print("Recall:", recall_score(y_test, y_pred, average='macro')*100, "%")

print("f1_score:", f1_score(y_test, y_pred, average='macro')*100 , "%")

print("Confusion-Matrix:", confusion_matrix(y_test,y_pred))
result = pd.DataFrame({
    "Actual": y_test,
    "Prediction": y_pred
})

print(result)

print("Classification-Report:", classification_report(y_test, y_pred))
```

```python
# Classification-Model --->  K-Nearest-Neghbour (KNN)

KNN_classifier= KNeighborsClassifier()

# use sklearn pipeline+gridsearch for make validation data unseen ()

from sklearn.pipeline import Pipeline
from  sklearn.model_selection import GridSearchCV

pipeline = Pipeline([
    ('scaler', StandardScaler()),  # only scale training data not validation data
    ('knn', KNeighborsClassifier())
])
K_val= {"knn__n_neighbors": [3,5,7,9]}
classifierCV= GridSearchCV(
    pipeline ,
    K_val ,    
    cv=5  ,   # folds
    scoring= "recall_macro"  # default accuracy
)

classifierCV.fit(x_train,y_train)
y_pred= classifierCV.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred)*100, "%")

print("Precision:",  precision_score(y_test, y_pred, average='macro')*100, "%")

print("Recall:", recall_score(y_test, y_pred, average='macro')*100, "%")

print("f1_score:", f1_score(y_test, y_pred, average='macro')*100 , "%")

print("Confusion-Matrix:", confusion_matrix(y_test,y_pred))
result = pd.DataFrame({
    "Actual": y_test,
    "Prediction": y_pred
})

print(result)

print("Classification-Report:", classification_report(y_test, y_pred))
```
